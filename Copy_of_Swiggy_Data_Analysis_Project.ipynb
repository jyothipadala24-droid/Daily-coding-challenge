```python
plt.figure(figsize=(14,6))

orange_shades = [
    '#FC8019',
    '#FF9F45',
    '#E65100',
    '#FFB266',
    '#CC5500',
    '#FF8C42',
    '#D2691E',
    '#FFA64D',
    '#C75100',
    '#FF7F24'
]

bars = plt.bar(
    Restaurant['name'].head(10),
    Restaurant['rating'].head(10),
    color=orange_shades,
    edgecolor='black'
)

plt.title(
    "Top 10 Restaurant Ratings",
    fontsize=16,
    fontweight='bold'
)

plt.xlabel("Restaurant Names", fontsize=12)

plt.ylabel("Ratings", fontsize=12)

plt.xticks(rotation=45)

for bar in bars:
    yval = bar.get_height()

    plt.text(
        bar.get_x() + bar.get_width()/2,
        yval,
        round(yval,1),
        ha='center',
        va='bottom',
        fontsize=11
    )

plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(8,5))

bars = Orders_Type['Type'].value_counts().plot(
    kind='bar',
    color=['#FC8019', '#C75100']
)

plt.title(
    "Swiggy Order Types",
    fontsize=15,
    fontweight='bold'
)

plt.xlabel("Order Type")

plt.ylabel("Total Orders")

plt.xticks(rotation=0)

for container in bars.containers:
    bars.bar_label(
        container,
        fontsize=12
    )

plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.show()
```

```python
Food['item'].value_counts().head(10)
```

```python
top_foods = Food['item'].value_counts().head(5)

plt.figure(figsize=(8,8))

plt.pie(
    top_foods.values,
    labels=top_foods.index,
    autopct='%1.1f%%',
    colors=['#FC8019', '#FF9F45', '#E65100', '#FFB266', '#CC5500'],
    shadow=True,
    startangle=90,
    explode=[0.03]*5
)

plt.title(
    "Top 5 Most Ordered Foods",
    fontsize=16,
    fontweight='bold'
)

plt.show()
```

```python
food_type = Food['veg_or_non_veg'].value_counts()

plt.figure(figsize=(8,8))

plt.pie(
    food_type.values,
    labels=food_type.index,
    autopct='%1.1f%%',
    colors=['#FC8019', '#C75100'],
    explode=[0.05, 0],
    shadow=True,
    startangle=90,
    textprops={'fontsize':14}
)

plt.title(
    "Veg vs Non-Veg Food Distribution",
    fontsize=18,
    fontweight='bold'
)

plt.legend(
    title="Food Type",
    loc="upper right"
)

plt.tight_layout()

plt.show()
```

```python
gender_count = Users['Gender'].value_counts()

plt.figure(figsize=(7,7))

plt.pie(
    gender_count.values,
    labels=gender_count.index,
    autopct='%1.1f%%',
    colors=['#FC8019', '#C75100'],
    shadow=True,
    startangle=90,
    explode=[0.05,0]
)

plt.title(
    "Gender Distribution",
    fontsize=18,
    fontweight='bold'
)

plt.show()
```

```python
occupation = Users['Occupation'].value_counts().head(5)

plt.figure(figsize=(10,5))

orange_shades = [
    '#FC8019',
    '#FF9F45',
    '#E65100',
    '#FFB266',
    '#CC5500'
]

bars = plt.bar(
    occupation.index,
    occupation.values,
    color=orange_shades,
    edgecolor='black'
)

plt.title(
    "Top Occupations",
    fontsize=16,
    fontweight='bold'
)

plt.xlabel("Occupation")

plt.ylabel("Users Count")

plt.xticks(rotation=20)

for bar in bars:
    yval = bar.get_height()

    plt.text(
        bar.get_x() + bar.get_width()/2,
        yval,
        int(yval),
        ha='center',
        va='bottom'
    )

plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.hist(
    Users['Age'],
    bins=10,
    color='#FC8019',
    edgecolor='black'
)

plt.title(
    "Age Distribution of Users",
    fontsize=16,
    fontweight='bold'
)

plt.xlabel("Age")

plt.ylabel("Number of Users")

plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.show()
```

```python
marital = Users['Marital Status'].value_counts()

plt.figure(figsize=(7,7))

plt.pie(
    marital.values,
    labels=marital.index,
    autopct='%1.1f%%',
    colors=['#FC8019','#FF9F45','#C75100'],
    shadow=True,
    startangle=90
)

plt.title(
    "Marital Status Distribution",
    fontsize=16,
    fontweight='bold'
)

plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.hist(
    Menu['price'],
    bins=10,
    color='#FC8019',
    edgecolor='black'
)

plt.title(
    "Menu Price Distribution",
    fontsize=16,
    fontweight='bold'
)

plt.xlabel("Price")

plt.ylabel("Number of Foods")

plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.show()
```

```python
top_prices = Menu.sort_values(
    by='price',
    ascending=False
).head(10)

plt.figure(figsize=(12,6))

orange_shades = [
    '#FC8019',
    '#FF9F45',
    '#E65100',
    '#FFB266',
    '#CC5500',
    '#FF8C42',
    '#D2691E',
    '#FFA64D',
    '#C75100',
    '#FF7F24'
]

bars = plt.bar(
    top_prices.index.astype(str),
    top_prices['price'],
    color=orange_shades,
    edgecolor='black'
)

plt.title(
    "Top 10 Highest Menu Prices",
    fontsize=16,
    fontweight='bold'
)

plt.xlabel("Menu Items")

plt.ylabel("Price")

for bar in bars:
    yval = bar.get_height()

    plt.text(
        bar.get_x() + bar.get_width()/2,
        yval,
        int(yval),
        ha='center',
        va='bottom'
    )

plt.show()
```

```python
price_groups = Menu['price'].head(5)

plt.figure(figsize=(8,8))

plt.pie(
    price_groups.values,
    labels=price_groups.index,
    autopct='%1.1f%%',
    colors=[
        '#FC8019',
        '#FF9F45',
        '#E65100',
        '#FFB266',
        '#CC5500'
    ],
    shadow=True,
    startangle=90
)

plt.title(
    "Menu Price Share",
    fontsize=16,
    fontweight='bold'
)

plt.show()
```

```python
order_type = Orders_Type['Type'].value_counts()

plt.figure(figsize=(9,9))

colors = ['#FC8019', '#FF9F45', '#C75100']

wedges, texts, autotexts = plt.pie(
    order_type.values,
    labels=order_type.index,
    autopct='%1.2f%%',
    colors=colors,
    startangle=90,
    shadow=True,
    explode=[0.03]*len(order_type),
    textprops={'fontsize':13}
)

centre_circle = plt.Circle((0,0),0.70,fc='white')

fig = plt.gcf()

fig.gca().add_artist(centre_circle)

plt.title(
    "Swiggy Order Types Distribution",
    fontsize=20,
    fontweight='bold'
)

plt.legend(
    wedges,
    [f'{label} : {value}' for label, value in zip(order_type.index, order_type.values)],
    title="Order Types",
    loc='upper right'
)

plt.tight_layout()

plt.show()
```
